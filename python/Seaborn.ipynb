# Seaborn

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/content/StudentsPerformance.csv')
data.head()
```

```python
sns.barplot(x='math score', y='race/ethnicity', data=data, orient='h')
```

### Agrupamos 'race/ethnicity' y promediamos  math score

```python
group_data = data.groupby('race/ethnicity')['math score'].mean().sort_values(ascending=False)
group_data
plt.barh(group_data.index, group_data.values, color='skyblue')
plt.show
```

### Promedio de calificaciones por matematica y tipo de lunch de los estudiantes

```python
sns.barplot(x='math score', y='race/ethnicity', hue='lunch', data=data, orient='h')
```

### Grafica con regresion del comportamiento de los datos

### Vemos una grafica de regresion de las calificaciones en lectura en la variable 'x' y en escritura en la variable 'y'

```python
ax = sns.regplot(x='reading score', y='writing score', data=data)
ax.set_title('Regresion Lineal')
```

### Ahora vemos la relacion que hay entre lectura y matematica

```python
ax = sns.regplot(x='reading score', y='math score', data=data)
ax.set_title('Regresion Lineal')
```

### Por ultimo vemos la relacion entre escritura y matematica

```python
ax = sns.regplot(x='writing score', y='math score', data=data)
ax.set_title('Regresion Lineal')
```

#### Graficos en una sola figura

```python
sns.pairplot(data, diag_kind ='kde', plot_kws={'alpha': 0.6, 's': 80, 'edgecolor': 'k'})
```

```python
sns.catplot(x='race/ethnicity', y='math score', hue='lunch', col='gender', kind='bar', data=data)
```
