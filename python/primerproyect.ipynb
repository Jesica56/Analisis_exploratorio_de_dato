<h3>Agregando columna Promedio</h3>

```python
data['promedio'] = data[['math score', 'reading score', 'writing score']].mean(axis=1).round(2) # creamos una nueva columna con el promedio de los puntajes
```

```python
data.head() 
```

<h5>Definiendo categoria_clasificacion </h5>

```python
def categoria_calificacion(promedio):
    if promedio >= 90:
        return 'A'
    elif promedio >= 80:
        return 'B'
    elif promedio >= 70:
        return 'C'
    elif promedio >= 60:
        return 'D'
    else:
        return 'F'
```

<h5>Agregando columna categoria_clasificacion</h5>

```python
data['categoria_promedio'] = data['promedio'].apply(categoria_calificacion) # creamos una nueva columna con la categoria de calificacion segun el promedio
```

```python
data.head()
```

<h3>Clasificacion de alumnos por categoria</h3>

```python
data['categoria_promedio'].value_counts() # contamos la cantidad de alumnos por categoria de calificacion
```

<h3>Correlacion entre puntajes</h3>

```python
data[['math score', 'reading score', 'writing score']].corr() # obtenemos la correlacion entre los puntajes
```

<h3>Vemos lo siguientes casos:</h3>
<ul>
<li>En matematica vemos q tiene una mayor puntaje, q estaria afectando a los puntajes de  lectura con 0.81 y escritura con 0.80</li>
 <li>En Lectura vemos q el puntaje es mayor q matematica con 0.81 cosa q estaria afectando a esta materia pero no asi a la materia de escritura con 0.95 de puntaje de correlacion</li>
 <li>En escritura vamos a ver q cuando el puntaje es mayor q a matematica afecta a esta materia, pero no asi a la materia de lectura.</li>
 </ul>

<h3>Agrupacion de datos por genero con (.groupby)</h3>

```python
data.groupby('gender')[['math score', 'reading score', 'writing score']].mean() # obtenemos el promedio de los puntajes por genero
```

<h3>Al agrupar por genero:</h3>
<p>
 Se puede ver que las mujeres obtienen mejores promedios en Lectura y Escritura que los hombre.
 <br>Los hombres obtienen mejores promedios en Matematica que las mujeres.</p>

<h3>Agrupacion del nivel educativo de los padres</h3>

```python
data.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean() # obtenemos el promedio de los puntajes por nivel educativo de los padres
```

<h3>Notamos que:</h3>
<ul>
<li>Los alumnos que tienen mejores notas en Matematica, Lectura y Escritura tienen padres con una maestria (master's degree)</li><br>
<li>Los alumnos que tienen notas bajas en Matematica, Lectura y Escritura tienen padres con escuela secundaria (High school)</li>
</ul>
<p>Basandonos en este conjunto de datos agrupados vemos que el nivel educativo de los padres si tiene relacion con la educacion de los hijos</p>

<h3>Agrupacion por almuerzo que reciben los estudiantes</h3>

```python
data.groupby('lunch')[['math score', 'reading score', 'writing score']].mean() # obtenemos el promedio de los puntajes por tipo de almuerzo
```

<p>Al finalizar este analisis se concluye que los estudiantes que obtuvieron los mejores promedios en Matematica, Lectura y Escritura tienden a ser aquellos, cuyos padres tienen un nivel educativo superior, como una maestria, y que ademas consumen almuerzos con mayor valor nutricional.<br>
Por otro lado, se observa que un buen desempeño en una materia puedo o no influir  de manera positiva en el rendimiento las demas, tanto en estudiantes hombres como en mujeres.
