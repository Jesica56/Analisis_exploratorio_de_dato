Student Performance in Exams
Este conjunto de datos contiene información sobre el rendimiento de distintos estudiantes en tres exámenes y algunas características demográficas. El conjunto tiene las siguientes columnas:

gender:
Indica el género del estudiante (hombre o mujer)
race/ethnicity:
Agrupa a los estudiantes en cinco categorías
parental level of education:
Muesta el nivel educativo más alto alcanzado por los padres del estudiante
some high school:
Significa que el padre/madre asistió a la secundaria pero no la terminó.
high school: 
Terminaron la secundaria pero no continuaron con sus estudios
some college: 
Asistieron a la universidad o una institución de educación superior, pero no la completo
associate degree: 
Es un título técnico o universitario de dos años (común en EE.UU.)
bachelor degree: 
Alguno de los padres obtuvo un título universitario
master degree: 
Alguno de los títulos obtuvo un título de posgrado (máster o maestría)
lunch:
Indica el tipo de almuerzo que recibe el estudiante
standard: 
Almuerzo de pago estándar
free/reduced: 
Almuerzo gratuito o con descuento
test preparation course:
Indica si el estudiante completó un curso de preparación antes del examen
math score:
Puntaje obtenido por el estudiante en matemáticas
reading score:
Puntaje obtenido por el estudiante en lectura
writing score:
Puntaje obtenido por el estudiante en escritura

```python
data['promedio'] = data[['math score', 'reading score', 'writing score']].mean(axis=1).round(2) # creamos una nueva columna con el promedio de los puntajes
```

```python
data.head() 
```

```python
def categoria_calificacion(promedio):
    if promedio >= 90:
        return 'A'
    elif promedio >= 80:
        return 'B'
    elif promedio >= 70:
        return 'C'
    elif promedio >= 60:
        return 'D'
    else:
        return 'F'
```

```python
data['categoria_promedio'] = data['promedio'].apply(categoria_calificacion) # creamos una nueva columna con la categoria de calificacion segun el promedio
```
