### NOTAS DE LOS ALUMNOS

```python
for alumno in alumnos:
    materias_str = ", ".join(f"{materia}: {nota}" for materia, nota in alumno["materias"].items())
    print(f"Alumno: {alumno['nombre']}")
    print(f"Materias: {materias_str}\n")
```

### Promedio de notas

```python
for alumno in alumnos:
    materias = alumno.get("materias", {})
    if materias:
        promedio = sum(materias.values()) / len(materias)
        porcentaje_notas = (promedio / 10) * 100
    else:
        porcentaje_notas = 0
        
        
    print(f"Nombre: {alumno['nombre']}")
    print(f"Materias: {materias_str}")
    print(f"Porcentaje promedio de notas: {porcentaje_notas:.2f}")
    print(f"Curso:", alumno["curso"])
    print()
```

### ASISTENCIA A CLASES Y PORCENTAJE

```python
for alumno in alumnos:
    total = alumno["asistencia"]["total_clases"]
    asistencias = alumno["asistencia"]["asistencias"]
    porcentaje_asistencia = (asistencias / total) * 100

    print(f"Nombre:", alumno["nombre"])
    print(f"Asistencia: {asistencias} de {total} clases")
    print(f"Porcentaje de asistencia: {porcentaje_asistencia:.2f}%")
    print(f"Curso:", alumno["curso"])
    print()
```
