Series apartir de un diccionarios

```python
data = {'informatica': 10, 'salud': 32, 'ingenieria': 53, 'humanidades': 74, 'economicas': 95} # creamos un diccionario
data # nos muestra el diccionario creado
```

```python
obj3 = pd.Series(data) # creamos un objeto de tipo Series a partir del diccionario
obj3 # nos muestra el objeto creado con sus indices y valores
```

```python
obj3.to_dict() # convertimos el objeto a un diccionario
obj3
```

```python
estudiantes = ['informatica', 'ingenieria', 'salud', 'arquitectura', 'economicas'] # creamos una lista de estudiantes
obj4 = pd.Series(data, index=estudiantes) # creamos un objeto de tipo Series a partir de la lista de estudiantes
obj4 # nos muestra el objeto creado con sus indices y valores
```

```python
pd.isna(obj4) # verificamos si el objeto es una Serie de pandas
```

```python
pd.notna(obj4) # verificamos si el objeto no es una Serie de pandas
```

```python
pd.isna(obj4).sum() # contamos los valores nulos en el objeto
```

```python
pd.notna(obj4).sum() # contamos los valores no nulos en el objeto
```

DATAFRAMES
Una forma de representar y trabajar con datos tabulares.
Diferencia de entre un dataFrame y una serie
El data frama puede manejar tipos de datos mixtos, como numeros, cadenas, booleanos o fechas, mientras que una Serie solo puede manejar un tipo de datos a la vez.

```python
data = {
    'estados': ['Aguascalientes', 'Baja California', 'Baja California Sur', 'Campeche', 'Chiapas'],
    'años': [2020, 2021, 2022, 2023, 2024],
    'poblacion': [1434635, 3769020, 798447, 928363, 554382],
    'superficie': [5616, 71222, 62215, 57000, 55000]
}
data
```

```python
frame = pd.DataFrame(data) # creamos un DataFrame a partir del diccionario
frame # nos muestra el DataFrame creado con sus indices y valores
```

```python
frame.head(3) # nos muestra las primeras 3 filas del DataFrame
```

```python
frame.tail(2) # nos muestra las ultimas 2 filas del DataFrame
```
