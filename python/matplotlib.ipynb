### Creacion de un grafico
#### se crea una figura con dos culumnas con graficos

```python
fig, axes = plt.subplots(2,2, sharex=True, sharey=True)

for  i in range(2):
  for j in range(2):
    axes[i,j].hist(np.random.standard_normal(500), bins=50, color='gray')
```

### Creacion de grafico a detalle

```python
plt.plot(np.random.standard_normal(50).cumsum(), color='green', linestyle='dashed', marker ='o')
```

```python
fig = plt.figure()
ax = fig.add_subplot()

data = np.random.standard_normal(50).cumsum()
ax.plot(data, color='black', linestyle='dashed')
ax.plot(data, color='blue', linestyle='dashed', drawstyle='steps-post')
```

## Modificacion de grafico


1.   Leyendas
2.   Etiquetas y mas



```python
fig, ax = plt.subplots()
ax.plot(np.random.standard_normal(1000).cumsum(), color= 'blue')
```

```python
fig, ax = plt.subplots()
ax.plot(np.random.standard_normal(1000).cumsum(), color= 'blue')


ticks = ax.set_xticks([0, 250, 500, 750, 1000])
labels = ax.set_xticklabels(['one', 'two', 'three', 'four', 'five'], rotation=30, fontsize='small')


ax.set_xlabel('Eje X')
ax.set_ylabel('Eje Y')
ax.set_title('Grafica')

plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(np.random.standard_normal(1000).cumsum(), color='gray', label='one')
ax.plot(np.random.standard_normal(1000).cumsum(), color='black', label='two', linestyle='dashed')
ax.plot(np.random.standard_normal(1000).cumsum(), color='blue', label='three', linestyle='dotted')

ax.legend()

plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(np.random.standard_normal(1000).cumsum(), color='gray', label='one')
ax.plot(np.random.standard_normal(1000).cumsum(), color='black', label='two', linestyle='dashed')
ax.plot(np.random.standard_normal(1000).cumsum(), color='blue', label='three', linestyle='dotted')

ticks = ax.set_xticks([0, 250, 500, 750, 1000])
labels = ax.set_xticklabels(['one', 'two', 'three', 'four', 'five'], rotation=30, fontsize='small')


ax.set_xlabel('Eje X')
ax.set_ylabel('Eje Y')
ax.set_title('Grafica')

ax.legend()

plt.show()
```

```python
fig, ax = plt.subplots(2,1)

data = pd.Series(np.random.uniform(size=16), index= list('abcdefghijklmnop'))

data.plot.bar(ax=ax[0], color='black', alpha=0.7)
data.plot.barh(ax=ax[1], color='black', alpha=0.7)
```
