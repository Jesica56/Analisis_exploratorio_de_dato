```python
import pandas as pd
import numpy as np
```

<h2>Manejo de datos faltante</h2>

```python
data = pd.Series(['a', 'b', np.nan, 'c', 'd', np.nan, 'e'])
data
```

```python
data.isna() #muestra si hay datos faltantes
```

<h3>Filtrar datos faltantes </h3>

```python
datos = pd.Series([1, np.nan, 3, 4 , np.nan, 6, 7, np.nan])
datos
```

```python
datos.dropna() #elimina los datos faltantes
```

```python
datos.notna() #muestra los datos que no son faltantes
```

```python
datos[datos.notna()] #filtra los datos que no son faltantes
```
