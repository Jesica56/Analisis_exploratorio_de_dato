<h3>Joins</h3>

```python
dta1 = pd.DataFrame({
    'key': ['b', 'b', 'a', 'c', 'a', 'a', 'd'], 'data1': range(7)
})
dta2 = pd.DataFrame({
    'key': ['a', 'b', 'c', 'd'], 'data2': range(4)
})
```

```python
dta1.head() #muestra las primeras filas del dataframe dta1
```

```python
dta2.head() #muestra las primeras filas del dataframe dta2
```

```python
pd.merge(dta1, dta2) #realiza un join por defecto, que es un inner join
```

```python
pd.merge(dta1, dta2, on= 'key') #realiza un join por la columna 'key'
```

```python
dta1 = pd.DataFrame({
    'lkey': ['b', 'b', 'a', 'c', 'a', 'a', 'd'], 'data1': range(7)
})
dta2 = pd.DataFrame({
    'rkey': ['a', 'b', 'c', 'd'], 'data2': range(4)
})
```

```python
dta1
```

```python
dta2
```

```python
pd.merge(dta1, dta2, left_on='lkey', right_on='rkey') #realiza un join por las columnas 'lkey' y 'rkey'
```

```python
pd.merge(dta1, dta2, left_on='lkey', right_on='rkey', how='outer') #realiza un join por las columnas 'lkey' y 'rkey', incluyendo todos los datos de ambos dataframes
```

```python
pd.merge(dta1, dta2, left_on='lkey', right_on='rkey', how='left') #realiza un join por las columnas 'lkey' y 'rkey', incluyendo todos los datos del dataframe dta1
```

```python
pd.merge(dta1, dta2, left_on='lkey', right_on='rkey' , how='right') #realiza un join por las columnas 'lkey' y 'rkey', incluyendo todos los datos del dataframe dta2
```

```python
pd.merge(dta1, dta2, left_on='lkey', right_on='rkey' , how='inner') #realiza un join por las columnas 'lkey' y 'rkey', incluyendo solo los datos que coinciden en ambos dataframes
```

<h3> Merge usando indices</h3>

```python
dta1 = pd.DataFrame({
    'key': ['b', 'b', 'a', 'c', 'a', 'a', 'd'], 'data1': range(7)
})
dta2 = pd.DataFrame({
    'group_val': [5, 6.7]
}, index = ['a', 'b']) #asigna los indices 'a' y 'b' a las filas del dataframe dta2
```

```python
dta1.head() #muestra las primeras filas del dataframe dta1
```

```python
dta2.head
```

```python
pd.merge(dta1, dta2, left_on='key', right_index=True) #realiza un join usando la columna 'key' del dataframe dta1 y el indice del dataframe dta2 
```

```python
pd.merge(dta1, dta2, left_on='key', right_index=True, how='outer') #realiza un join por las columnas 'key'  incluyendo todos los datos de ambos dataframes y muestra de dónde proviene cada fila
```

```python
pd.merge(dta1, dta2, left_on='key')
```
