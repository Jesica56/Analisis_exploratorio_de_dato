## Analisamos de datos CO2 per capita con mayor y menor emision 

```python
ultimo_año = info['Year'].max()
```

```python
info_ultimo = info[info['Year'] == ultimo_año]
```

```python
mayor = info_ultimo['Annual CO₂ emissions (per capita)'].max()
info_ultimo[info_ultimo['Annual CO₂ emissions (per capita)'] == mayor]
```

```python
menor = info_ultimo['Annual CO₂ emissions (per capita)'].min()
info_ultimo[info_ultimo['Annual CO₂ emissions (per capita)'] == menor]
```

### Agrupamos los paises con minimo y maximo historicos

```python
maximos = info.groupby('Entity')['Annual CO₂ emissions (per capita)'].max().sort_values(ascending=False)
maximos.head(20)
```

```python
minimo = info.groupby('Entity')['Annual CO₂ emissions (per capita)'].min().sort_values()
minimo.head(20)
```
